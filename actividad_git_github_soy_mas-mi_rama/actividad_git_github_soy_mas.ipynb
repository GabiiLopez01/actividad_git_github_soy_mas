¡Bienvenidas!

Este repositorio está diseñado para que se pueda utilizar para la actividad que se va a presentar y de esta manera puedan contribuir a un proyecto en GitHub.
El objetivo principal es agregar una línea en este **README.md** para registrar tu contribución.

**Instrucciones básicas:**

1.   Clona este repositorio.
2.   Crea una nueva rama.
3.   Abre el archivo README.md y agrega una línea con tu nombre o mensaje de contribución.
4.   Haz commit de tus cambios y súbelos al repositorio remoto.

(Opcional) *Crear un Pull Request para integrar tus cambios a la rama principal (main)*.
